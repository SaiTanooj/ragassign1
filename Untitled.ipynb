```python
import time
import numpy as np
import faiss
from pypdf import PdfReader
from sentence_transformers import SentenceTransformer
```

```python
reader = PdfReader(r"C:\Users\Tanooj\Downloads\a202345.pdf")
full_text = " ".join(page.extract_text() or "" for page in reader.pages)
print(len(reader.pages), "pages,", len(full_text.split()), "words")
```

```python
def chunk_text(text, size=150, overlap=30):
    words = text.split()
    step = size - overlap
    return [" ".join(words[i:i + size]) for i in range(0, len(words), step)]

chunks = chunk_text(full_text)
print(len(chunks), "chunks")
print(chunks[0][:300])
```

```python
embedder = SentenceTransformer("all-MiniLM-L6-v2")   # the Bi-Encoder

embeddings = embedder.encode(chunks, normalize_embeddings=True, show_progress_bar=True)
embeddings = np.array(embeddings).astype("float32")

index = faiss.IndexFlatIP(embeddings.shape[1])   # inner product
index.add(embeddings)
```

```python
def retrieve_top_k(query, k=10):
    q_emb = embedder.encode([query], normalize_embeddings=True).astype("float32")
    scores, ids = index.search(q_emb, k)
    return [(int(i), float(s)) for i, s in zip(ids[0], scores[0])]

query = "Explain laws for rioting"

start = time.perf_counter()
top10 = retrieve_top_k(query)
baseline_retrieval_time = time.perf_counter() - start

for rank, (idx, score) in enumerate(top10, 1):
    print(f"Rank {rank} | chunk {idx} | score {score:.3f}")
```

```python
def build_prompt(query, chunk_ids):
    context = "\n\n".join(f"[Chunk {i}]\n{chunks[i]}" for i in chunk_ids)
    return f"""Answer the question using ONLY the context below.
If the answer is not in the context, say "Not found in the context."

Context:
{context}

Question: {query}
Answer:"""

top3_ids = [idx for idx, _ in top10[:3]]
prompt = build_prompt(query, top3_ids)
```

```python
from sentence_transformers import CrossEncoder

reranker = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")
```

```python
def rerank(query, top10):
    ids = [idx for idx, _ in top10]
    pairs = [(query, chunks[i]) for i in ids]      # 10 (query, chunk) pairs

    start = time.perf_counter()
    scores = reranker.predict(pairs)
    rerank_time = time.perf_counter() - start

    ranked = sorted(zip(ids, scores), key=lambda x: x[1], reverse=True)
    return [(i, float(s)) for i, s in ranked], rerank_time
```

```python
def run_both(query, answer_phrase=None):
    # Pipeline 1: baseline Top 10
    t0 = time.perf_counter()
    top10 = retrieve_top_k(query)
    base_time = time.perf_counter() - t0

    # Pipeline 2: re-rank the same 10
    reranked, rr_time = rerank(query, top10)

    old_rank = {cid: r for r, (cid, _) in enumerate(top10, 1)}

    print("QUERY:", query)
    print(f"Baseline: {base_time:.3f}s | Re-rank: {rr_time:.3f}s\n")
    for new_r, (cid, score) in enumerate(reranked, 1):
        star = "★" if answer_phrase and answer_phrase.lower() in chunks[cid].lower() else " "
        print(f"{star} New #{new_r:<2} (was #{old_rank[cid]:<2}) | chunk {cid:<4} | {chunks[cid][:70]}...")
        if new_r == 3:
            print("-------- only the top 3 go to the LLM --------")

    return top10, reranked, base_time, rr_time
```

```python
q = "Is killing always murder?"
phrase = "self-control by grave and sudden provocation, causes the death"

top10, reranked, base_time, rr_time = run_both(q, phrase)
```

```python
answer_1 = ask_llm(build_prompt(q, [i for i, _ in top10[:3]]))        # baseline top 3
answer_2 = ask_llm(build_prompt(q, [i for i, _ in reranked[:3]]))     # re-ranked top 3

print("ANSWER 1 (baseline):\n", answer_1)
print("\nANSWER 2 (re-ranked):\n", answer_2)
```

```python
def latency(query, runs=5):
    base, rr = [], []
    for _ in range(runs):
        t0 = time.perf_counter()
        t10 = retrieve_top_k(query)
        base.append(time.perf_counter() - t0)
        rr.append(rerank(query, t10)[1])
    b, r = sum(base) / runs, sum(rr) / runs
    print(f"Pipeline 1 (baseline):          {b:.3f}s")
    print(f"Pipeline 2 (baseline + rerank): {b + r:.3f}s")
    print(f"Re-ranking adds {r:.3f}s (~{r / b:.0f}x the baseline time)")

latency(q)
```

```python
def deep_rank(query, good, k=100):
    """Baseline rank of the answer chunk among the top k (not just the top 10)."""
    q = embedder.encode([query], normalize_embeddings=True).astype("float32")
    _, ids = index.search(q, k)
    for r, cid in enumerate(ids[0], 1):
        if int(cid) in good:
            return r
    return None


def rank_check(tests):
    fmt = lambda r: f"#{r}" if r else "-"
    print(f"{'TYPE':<14}{'BASELINE':<10}{'RE-RANKED':<11}{'TRUE RANK':<11}{'VERDICT':<11}QUERY")
    for kind, q, phrase in tests:
        good = {i for i, c in enumerate(chunks) if phrase.lower() in c.lower()}
        if not good:
            print(f"{kind:<14}⚠️ phrase not found in any chunk: {phrase[:50]}")
            continue

        top10 = retrieve_top_k(q)
        reranked, _ = rerank(q, top10)
        b = next((r for r, (c, _) in enumerate(top10, 1) if c in good), None)
        n = next((r for r, (c, _) in enumerate(reranked, 1) if c in good), None)
        deep = deep_rank(q, good)

        if (b is None or b > 3) and n is not None and n <= 3: v = "FIXED"
        elif b is not None and b <= 3 and (n is None or n > 3): v = "HURT"
        elif b is None: v = "MISS"
        elif b > 3: v = "BOTH MISS"
        else: v = "OK"
        print(f"{kind:<14}{fmt(b):<10}{fmt(n):<11}{fmt(deep):<11}{v:<11}{q[:55]}")


tests = [
    # Already confirmed earlier
    ("Exception", "Is killing always murder?",
     "self-control by grave and sudden provocation, causes the death"),
    ("Exception", "Is it murder if the victim agreed to die?",
     "suffers death or takes the risk of death with his own consent"),
    ("Negation", "Can someone who was drunk when committing a crime escape punishment?",
     "administered to him without his knowledge"),

    # Keyword trap / negation attempts
    ("Keyword trap", "What is the punishment for breaking into someone's house at night?",
     "after sunset and before sunrise, shall be punished"),
    ("Keyword trap", "What is the punishment for stealing a child?",
     "kidnaps any person from India or from lawful guardianship shall be punished"),
    ("Negation", "Can I hit a police officer who is arresting me?",
     "colour of his office, though that act, may not be strictly justifiable by law"),
    ("Negation", "Is there a limit on how much force I can use to defend myself?",
     "in no case extends to the inflicting of more harm"),
    ("Keyword trap", "Is it a crime to keep money I found on the road?",
     "Whoever dishonestly misappropriates or converts to his own use any movable property, shall be punished"),
]

tests2 = [  # closer wording, between everyday and legal language (untested)
    ("Negation", "Is there any right of self-defence against a public officer doing his duty?",
     "colour of his office, though that act, may not be strictly justifiable by law"),
    ("Negation", "Does the right of self-defence allow causing more harm than necessary?",
     "in no case extends to the inflicting of more harm"),
    ("Keyword trap", "What is the punishment for taking a child away from the parents without their permission?",
     "kidnaps any person from India or from lawful guardianship shall be punished"),
]

rank_check(tests + tests2)
```
